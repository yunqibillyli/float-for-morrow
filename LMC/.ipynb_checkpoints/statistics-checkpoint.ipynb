```python
import numpy as np
```

```python
def random_effects(means, stds):

    means = np.array(means)
    stds = np.array(stds)
    
    var  = stds ** 2
    w_FE = 1 / var
    mu_FE = (w_FE * means).sum() / w_FE.sum()

    Q  = (w_FE * (means - mu_FE) ** 2).sum()
    df = len(means) - 1
    c  = w_FE.sum() - (w_FE ** 2).sum() / w_FE.sum()

    T_squared = max(0.0, (Q - df) / c)

    weights = 1 / (var + T_squared)

    combined_mean = (weights * means).sum() / weights.sum()
    combined_std  = np.sqrt(1 / weights.sum())

    return combined_mean, combined_std
```

```python
# 30 Doradus Oxygen
means = np.array([8.275000000000002, # BAT-99-105, E140M lower res, DLA reliability B
                  7.149999999999999, # RMC-140, likely false detection, large uncertainty, DLA reliability B
                  # not detected for VFTS-440
                  # not detected for VFTS-545
                  8.48, # VFTS-482, weak detection, DLA reliability B
                  8.331, # VFTS-542, weak detection, DLA reliability A
                  # not detected for VFTS-506
                  7.300000000000001 # VFTS-406, likely false detection, DLA reliability A
                  # not detected for VFTS-285
                 ])
stds = np.array([0.21972937900972642, 
                 0.8557008823181147, # RMC-140
                 0.20270421801235416, 
                 0.20909567188251413, 
                 0.6667053322120651 # VFTS-406
                ])
random_effects(means, stds)
```
