```python
O_30_doradus_path = "/Users/billyli/Documents/cloudy/float-for-morrow/30_doradus.ele_O"
Fe_30_doradus_path = "/Users/billyli/Documents/cloudy/float-for-morrow/30_doradus.ele_Fe"
O_n66a_path = "/Users/billyli/Documents/cloudy/float-for-morrow/n66a.ele_O"
Fe_n66a_path = "/Users/billyli/Documents/cloudy/float-for-morrow/n66a.ele_Fe"
O_solar_path = "/Users/billyli/Documents/cloudy/float-for-morrow/solar.ele_O"
Fe_solar_path = "/Users/billyli/Documents/cloudy/float-for-morrow/solar.ele_Fe"
O_i_zw_18_path = "/Users/billyli/Documents/cloudy/float-for-morrow/i_zw_18.ele_O"
Fe_i_zw_18_path = "/Users/billyli/Documents/cloudy/float-for-morrow/i_zw_18.ele_Fe"

O_30_doradus = pd.read_csv(O_30_doradus_path, delim_whitespace = True, comment = "#", header = None)
Fe_30_doradus = pd.read_csv(Fe_30_doradus_path, delim_whitespace = True, comment = "#", header = None)
O_n66a = pd.read_csv(O_n66a_path, delim_whitespace = True, comment = "#", header = None)
Fe_n66a = pd.read_csv(Fe_n66a_path, delim_whitespace = True, comment = "#", header = None)
O_solar = pd.read_csv(O_solar_path, delim_whitespace = True, comment = "#", header = None)
Fe_solar = pd.read_csv(Fe_solar_path, delim_whitespace = True, comment = "#", header = None)
O_i_zw_18 = pd.read_csv(O_i_zw_18_path, delim_whitespace = True, comment = "#", header = None)
Fe_i_zw_18 = pd.read_csv(Fe_i_zw_18_path, delim_whitespace = True, comment = "#", header = None)

def ratios(O_df, Fe_df):
    OII = O_df[2].astype(float)
    OIII = O_df[3].astype(float)
    FeII = Fe_df[2].astype(float)
    FeIII = Fe_df[3].astype(float)
    FeIV = Fe_df[4].astype(float)
    FeV = Fe_df[5].astype(float)
    with np.errstate(divide = 'ignore', invalid = 'ignore'):
        x = np.log10(OIII / OII)
        y = np.log10((FeII + FeIV + FeV) / FeIII)
    return x, y

x_30_doradus, y_30_doradus = ratios(O_30_doradus, Fe_30_doradus)
x_n66a, y_n66a = ratios(O_n66a, Fe_n66a)
x_solar, y_solar = ratios(O_solar, Fe_solar)
x_i_zw_18, y_i_zw_18 = ratios(O_i_zw_18, Fe_i_zw_18)

plt.figure(figsize = (6, 6), dpi = 100)

plt.scatter(x_solar, y_solar, marker = "o", color = "0", label = "Solar Metallicity (0)")
plt.scatter(x_30_doradus, y_30_doradus, marker = "o", color = "0.3", label = "LMC Metallicity (-0.5)")
plt.scatter(x_n66a, y_n66a, marker = "o", color = "0.6", label = "SMC Metallicity (-0.7)")
plt.scatter(x_i_zw_18, y_i_zw_18, marker = "o", color = "0.9", label = "I Zw 18 Metallicity (-1.5)")

x_array = np.linspace(0.1, 1.5, 100)
y_array = np.log10(0.9 * (10 ** -x_array) ** 0.08 + 0.9 * (10 ** -x_array) ** -0.92 - 1)

plt.plot(x_array, y_array, color = "red", linestyle = "--", linewidth = 3, label = "Rodriguez and Rubin (2005) Eq. 2")

x_array_2 = np.linspace(0.1, 1.35, 100)
y_array_2 = np.log10(1.1 * (10 ** -x_array_2) ** 0.58 + 1.1 * (10 ** -x_array_2) ** -0.42 - 1)

plt.plot(x_array_2, y_array_2, color = "red", linestyle = "-", linewidth = 3, label = "Rodriguez and Rubin (2005) Eq. 3 and 4")

x_array_3 = np.linspace(0.1, 1.35, 100)
v_array_3 = 1 / (1 + (10 ** x_array_3))
y_array_3 = np.log10((0.036 * v_array_3 - 0.146 + 1.386 / v_array_3) - 1)

plt.plot(x_array_3, y_array_3, color = "purple", linestyle = "--", linewidth = 3, label = r"Izotov+2006 12+log(O/H)$\leq$7.2")

x_array_4 = np.linspace(0.1, 1.35, 100)
v_array_4 = 1 / (1 + (10 ** x_array_4))
y_array_4 = np.log10((0.301 * v_array_4 - 0.259 + 1.367 / v_array_4) - 1)

plt.plot(x_array_4, y_array_4, color = "purple", linestyle = "-", linewidth = 3, label = r"Izotov+2006 7.2$\leq$12+log(O/H)$\leq$8.2")

x_array_5 = np.linspace(0.1, 1.35, 100)
v_array_5 = 1 / (1 + (10 ** x_array_5))
y_array_5 = np.log10((-1.377 * v_array_5 + 1.606 + 1.045 / v_array_5) - 1)

plt.plot(x_array_5, y_array_5, color = "purple", linestyle = "-.", linewidth = 3, label = r"Izotov+2006 12+log(O/H)$\geq$8.2")

plt.xlim(0.1, 1.35)
plt.ylim(0, 1.5)
plt.xlabel("log(O III/O II)", fontsize = 15)
plt.ylabel("log((Fe II + Fe IV + Fe V)/Fe III)", fontsize = 15)
plt.grid(True, linestyle = "--", alpha = 0.4)
plt.legend(loc = "best")
plt.tight_layout()

plt.savefig('Fe_ionization_corrections.png', dpi = 300)
plt.show()
```

```python
def ratios_FeII(O_df, Fe_df):
    OII = O_df[2].astype(float)
    OIII = O_df[3].astype(float)
    FeII = Fe_df[2].astype(float)
    FeIII = Fe_df[3].astype(float)
    FeIV = Fe_df[4].astype(float)
    FeV = Fe_df[5].astype(float)
    with np.errstate(divide = 'ignore', invalid = 'ignore'):
        x = np.log10(OIII / OII)
        y = np.log10(FeII / FeIII)
    return x, y

def ratios_FeIV(O_df, Fe_df):
    OII = O_df[2].astype(float)
    OIII = O_df[3].astype(float)
    FeII = Fe_df[2].astype(float)
    FeIII = Fe_df[3].astype(float)
    FeIV = Fe_df[4].astype(float)
    FeV = Fe_df[5].astype(float)
    with np.errstate(divide = 'ignore', invalid = 'ignore'):
        x = np.log10(OIII / OII)
        y = np.log10(FeIV / FeIII)
    return x, y

def ratios_FeV(O_df, Fe_df):
    OII = O_df[2].astype(float)
    OIII = O_df[3].astype(float)
    FeII = Fe_df[2].astype(float)
    FeIII = Fe_df[3].astype(float)
    FeIV = Fe_df[4].astype(float)
    FeV = Fe_df[5].astype(float)
    with np.errstate(divide = 'ignore', invalid = 'ignore'):
        x = np.log10(OIII / OII)
        y = np.log10(FeV / FeIII)
    return x, y

x_FeII, y_FeII = ratios_FeII(O_n66a, Fe_n66a)
x_FeIV, y_FeIV = ratios_FeIV(O_n66a, Fe_n66a)
x_FeV, y_FeV = ratios_FeV(O_n66a, Fe_n66a)

plt.figure(figsize = (8, 8), dpi = 100)

plt.scatter(x_FeII, y_FeII, marker = "o", color = "0.3", label = "log(Fe II/Fe III)")
plt.scatter(x_FeIV, y_FeIV, marker = "o", color = "0.6", label = "log(Fe IV/Fe III)")
plt.scatter(x_FeV, y_FeV, marker = "o", color = "0.9", label = "log(Fe V/Fe III)")

plt.xlabel("log(O III/O II)")
plt.ylabel("ICF")
plt.grid(True, linestyle = "--", alpha = 0.4)
plt.legend(loc = "best")
plt.tight_layout()
plt.show()
```
